# Homogenization scheme derivations

This appendix provides the formal derivations of the concentration equations for the main homogenization schemes implemented in *Echoes*. For a comprehensive treatment, see [@zaoui2002], [@bornert2001a], or [@nemat-nasser1999].

## General framework

Consider an RVE composed of $N$ phases with stiffness tensors $\mathbb{C}_r$ and volume fractions $f_r$. The effective stiffness $\mathbb{C}^{hom}$ is defined by

$$
\mathbb{C}^{hom} = \sum_{r=1}^{N} f_r \, \mathbb{C}_r : \mathbb{A}_r
$$

where $\mathbb{A}_r$ is the **strain concentration tensor** of phase $r$, satisfying

$$
\sum_{r=1}^{N} f_r \, \mathbb{A}_r = \mathbb{I}
$$

The key difference between schemes lies in how $\mathbb{A}_r$ is estimated.

## Voigt and Reuss bounds

The **Voigt** (uniform strain) bound assumes $\mathbb{A}_r = \mathbb{I}$ for all phases:

$$
\mathbb{C}^{Voigt} = \sum_{r=1}^{N} f_r \, \mathbb{C}_r
$$

The **Reuss** (uniform stress) bound assumes $\mathbb{A}_r = \mathbb{C}_r^{-1} : \left(\sum_s f_s \, \mathbb{C}_s^{-1}\right)^{-1}$, yielding:

$$
\mathbb{C}^{Reuss} = \left(\sum_{r=1}^{N} f_r \, \mathbb{C}_r^{-1}\right)^{-1}
$$

These are rigorous bounds for any composite with prescribed volume fractions.

## Dilute scheme (DIL)

In the dilute approximation, each inclusion is embedded in the matrix (phase 0) alone, without interaction with other inclusions. The concentration tensor is

$$
\mathbb{A}_r^{dil} = \left(\mathbb{I} + \mathbb{P}_0^r : \delta\mathbb{C}_r\right)^{-1}
$$

where $\mathbb{P}_0^r$ is the Hill polarization tensor of the inclusion $r$ in the matrix and $\delta\mathbb{C}_r = \mathbb{C}_r - \mathbb{C}_0$. The effective stiffness is

$$
\mathbb{C}^{DIL} = \mathbb{C}_0 + \sum_{r=1}^{N} f_r \, \delta\mathbb{C}_r : \mathbb{A}_r^{dil}
$$

This scheme is valid only at low concentrations since $\sum_r f_r \mathbb{A}_r^{dil} \neq \mathbb{I}$ in general.

## Mori-Tanaka scheme (MT)

The Mori-Tanaka scheme corrects the dilute estimate by enforcing the constraint ([eq]). The strain concentration tensor is

$$
\mathbb{A}_r^{MT} = \mathbb{A}_r^{dil} : \left(\sum_{s=0}^{N} f_s \, \mathbb{A}_s^{dil}\right)^{-1}
$$

with $\mathbb{A}_0^{dil} = \mathbb{I}$. This yields

$$
\mathbb{C}^{MT} = \left(\sum_{r=0}^{N} f_r \, \mathbb{C}_r : \mathbb{A}_r^{dil}\right) : \left(\sum_{s=0}^{N} f_s \, \mathbb{A}_s^{dil}\right)^{-1}
$$

The MT scheme is well-suited for matrix-inclusion composites with a well-defined connected matrix phase.

## Self-consistent scheme (SC)

In the self-consistent scheme, each inclusion is embedded in the (unknown) effective medium itself. The concentration tensor satisfies

$$
\mathbb{A}_r^{SC} = \left(\mathbb{I} + \mathbb{P}_{hom}^r : (\mathbb{C}_r - \mathbb{C}^{hom})\right)^{-1}
$$

where $\mathbb{P}_{hom}^r$ is the Hill tensor computed with the effective stiffness $\mathbb{C}^{hom}$. Since $\mathbb{C}^{hom}$ appears on both sides, this leads to an **implicit** (nonlinear) equation:

$$
\mathbb{C}^{SC} = \sum_{r=1}^{N} f_r \, \mathbb{C}_r : \mathbb{A}_r^{SC}
\qquad\text{with}\qquad
\sum_{r=1}^{N} f_r \, \mathbb{A}_r^{SC} = \mathbb{I}
$$

which is solved iteratively. The SC scheme naturally accounts for phase interactions and predicts percolation thresholds. It is particularly suited for polycrystalline or disordered materials where no phase plays the role of a connected matrix.

## Pont-Castañeda-Willis scheme (PCW)

The PCW scheme [@willis1977] introduces a **distribution** Hill tensor $\mathbb{P}_0^d$ (associated with a distribution ellipsoid that describes the spatial arrangement of inclusions) distinct from the **shape** Hill tensor $\mathbb{P}_0^r$. The effective stiffness is

$$
\mathbb{C}^{PCW} = \mathbb{C}_0 + \left[\left(\sum_{r=1}^{N} f_r \, \delta\mathbb{C}_r : \mathbb{A}_r^{dil}\right)^{-1} + \mathbb{P}_0^d\right]^{-1}
$$

When $\mathbb{P}_0^d = \mathbb{P}_0^r$ for all phases (same distribution and shape), this reduces to the MT scheme. The PCW scheme provides more flexibility to model non-uniform spatial distributions.

## Maxwell scheme (MAX)

The Maxwell scheme considers that all inclusions are enclosed in a large ellipsoidal domain (the "Maxwell domain") embedded in the matrix. Equating the far-field perturbation of this domain (treated as a single equivalent inhomogeneity) with the sum of individual perturbations yields

$$
\mathbb{C}^{MAX} = \mathbb{C}_0 + \left[\left(\sum_{r=1}^{N} f_r \, \delta\mathbb{C}_r : \mathbb{A}_r^{dil}\right)^{-1} + \mathbb{P}_0^{MAX}\right]^{-1}
$$

where $\mathbb{P}_0^{MAX}$ is the Hill tensor of the Maxwell domain. The result formally coincides with ([eq]) where the distribution tensor is that of the Maxwell domain.

## Hashin-Shtrikman bounds

The Hashin-Shtrikman (HS) bounds provide the tightest possible bounds for isotropic composites with prescribed volume fractions and isotropic phases. They correspond to specific Mori-Tanaka estimates:

- **HS lower bound ($\mathbb{C}^{HS-}$):** MT scheme with the softest phase as matrix
- **HS upper bound ($\mathbb{C}^{HS+}$):** MT scheme with the stiffest phase as matrix

For a two-phase isotropic composite with bulk/shear moduli $(k_1,\mu_1)$ and $(k_2,\mu_2)$ with $k_1 \leq k_2$ and $\mu_1 \leq \mu_2$:

$$
k^{HS-} = k_1 + \frac{f_2}{\frac{1}{k_2 - k_1} + \frac{3\,f_1}{3k_1+4\mu_1}}
\qquad
\mu^{HS-} = \mu_1 + \frac{f_2}{\frac{1}{\mu_2-\mu_1} + \frac{6(k_1+2\mu_1)\,f_1}{5\mu_1(3k_1+4\mu_1)}}
$$

## Differential scheme (DIFF)

The differential scheme builds up the composite incrementally. Starting from the matrix, a small volume fraction of inclusions is added at each step, and the effective properties are updated. This leads to the ordinary differential equation

$$
\frac{\,\mathrm{d}\mathbb{C}^{hom}(f)}{\,\mathrm{d} f} = \frac{1}{1-f} \, \delta\mathbb{C} : \left(\mathbb{I} + \mathbb{P}_{hom}^r(f) : \delta\mathbb{C}\right)^{-1}
$$

with initial condition $\mathbb{C}^{hom}(0) = \mathbb{C}_0$ and where $\delta\mathbb{C} = \mathbb{C}_1 - \mathbb{C}^{hom}(f)$. The ODE is integrated numerically up to the desired volume fraction. The differential scheme is particularly useful for high concentrations and naturally avoids the percolation artifacts of other schemes.

## Dilute stress scheme (DILD)

The DILD scheme is the **stress-based (dual) counterpart** of the DIL scheme. Instead of prescribing uniform strain in the reference medium, it prescribes uniform stress, working with compliance tensors $\mathbb{S}_r = \mathbb{C}_r^{-1}$. The dual Hill tensor is

$$
\mathbb{Q}_0 = \mathbb{C}_0 - \mathbb{C}_0 : \mathbb{P}_0 : \mathbb{C}_0
$$

The **stress concentration tensor** of inclusion $r$ in the dilute approximation is

$$
\mathbb{B}_r^{dil} = \left(\mathbb{I} + \mathbb{Q}_0 : \delta\mathbb{S}_r\right)^{-1}
$$

where $\delta\mathbb{S}_r = \mathbb{S}_r - \mathbb{S}_0$. The effective compliance is

$$
\mathbb{S}^{DILD} = \mathbb{S}_0 + \sum_{r \neq 0} f_r \, \delta\mathbb{S}_r : \mathbb{B}_r^{dil}
$$

Like the DIL scheme, DILD is valid only at low concentrations. While DIL overestimates stiffness, DILD overestimates compliance. Together they bracket the true effective properties at low volume fractions.

## Asymmetric self-consistent scheme (ASC)

The ASC scheme was introduced by @saevik2014 as an **asymmetric** variant of the self-consistent scheme. Unlike the symmetric SC scheme where all phases are treated equivalently, ASC preserves a distinction between the matrix (phase 0) and the inclusions: the matrix stiffness $\mathbb{C}_0$ enters directly, while the inclusions use the effective medium $\mathbb{C}^{ASC}$ as their reference medium.

The fixed-point equation is

$$
\mathbb{C}^{ASC} = \mathbb{C}_0 + \sum_{r \neq 0} f_r \left(\mathbb{C}_r - \mathbb{C}^{ASC}\right) : \left(\mathbb{I} + \mathbb{P}\!\left(\mathbb{C}^{ASC}, \boldsymbol{A}_r\right) : \left(\mathbb{C}_r - \mathbb{C}^{ASC}\right)\right)^{-1}
$$

where $\mathbb{P}(\mathbb{C}^{ASC}, \boldsymbol{A}_r)$ is the Hill polarization tensor of inclusion $r$ computed in the effective medium $\mathbb{C}^{ASC}$. Since $\mathbb{C}^{ASC}$ appears on both sides, [eq] is solved iteratively, starting from the MT estimate as initial guess.

Compared to SC, the ASC scheme:

- retains the matrix-inclusion topology (no percolation threshold artifact),
- accounts for the stiffness contrast relative to the effective medium rather than the matrix alone,
- is better suited than MT when inclusion concentrations are moderate and phase contrast is large.
