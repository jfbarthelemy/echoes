# Hill polarization tensors

The Hill polarization tensors defined in this section are related to an ellipsoid embedded in the whole space $\mathbb{R}^3$. An ellipsoid $\mathcal{E}_{\boldsymbol{A}}$ can be defined by an equation such as
$$
\underline{x}\in\mathcal{E}_{\boldsymbol{A}}
\quad\Leftrightarrow\quad
\underline{x}\cdot({}^{t}{\boldsymbol{A}}\cdot\boldsymbol{A})^{-1}\cdot\underline{x}\leq 1
$$
where $\boldsymbol{A}$ is an invertible second-order tensor so that ${}^{t}{\boldsymbol{A}}\cdot\boldsymbol{A}$ is a positive definite symmetric tensor associated to 3 radii (eigenvalues $a\geq b \geq c$ possibly written $\rho_1 \geq \rho_2 \geq \rho_3$ for convenience) and 3 angles (orientation of the frame of orthonormal eigenvectors $\underline{e}^{\boldsymbol{A}}_1, \underline{e}^{\boldsymbol{A}}_2, \underline{e}^{\boldsymbol{A}}_3$)
$$
{}^{t}{\boldsymbol{A}}\cdot\boldsymbol{A}=a^2 \underline{e}^{\boldsymbol{A}}_1\otimes\underline{e}^{\boldsymbol{A}}_1 + b^2 \underline{e}^{\boldsymbol{A}}_2\otimes\underline{e}^{\boldsymbol{A}}_2 + c^2 \underline{e}^{\boldsymbol{A}}_3\otimes\underline{e}^{\boldsymbol{A}}_3 = \sum_{i=1}^3 \rho_i^2 \underline{e}^{\boldsymbol{A}}_i\otimes\underline{e}^{\boldsymbol{A}}_i
$$

## Useful integrals

Some integrals useful for further calculations and intimately connected to the Newtonian potential are introduced as

$$\begin{align}
\boldsymbol{I}^{\boldsymbol{A}} &= \frac{\det{\boldsymbol{A}}}{4\pi}
\int_{\lVert \underline{\xi} \rVert=1}
\frac{\underline{\xi}\otimes\underline{\xi}}
{\lVert \boldsymbol{A}\cdot\underline{\xi} \rVert^3}\,\mathrm{d} S_{\xi}
=
\frac{1}{4\pi}
\int_{\lVert \underline{\zeta} \rVert=1}
\frac{%
(\boldsymbol{A}^{-1}\cdot\underline{\zeta})
\otimes
(\boldsymbol{A}^{-1}\cdot\underline{\zeta})
}{\lVert \boldsymbol{A}^{-1}\cdot\underline{\zeta} \rVert^2}
\,\mathrm{d} S_{\zeta}
\\
\mathbb{U}^{\boldsymbol{A}} &= \frac{\det{\boldsymbol{A}}}{4\pi}
\int_{\lVert \underline{\xi} \rVert=1}
\frac{\underline{\xi}\otimes\underline{\xi}\otimes\underline{\xi}\otimes\underline{\xi}}
{\lVert \boldsymbol{A}\cdot\underline{\xi} \rVert^3}\,\mathrm{d} S_{\xi}\\
&=
\frac{1}{4\pi}
\int_{\lVert \underline{\zeta} \rVert=1}
\frac{%
(\boldsymbol{A}^{-1}\cdot\underline{\zeta})
\otimes
(\boldsymbol{A}^{-1}\cdot\underline{\zeta})
\otimes
(\boldsymbol{A}^{-1}\cdot\underline{\zeta})
\otimes
(\boldsymbol{A}^{-1}\cdot\underline{\zeta})
}{\lVert \boldsymbol{A}^{-1}\cdot\underline{\zeta} \rVert^4}
\,\mathrm{d} S_{\zeta}
\\
\mathbb{V}^{\boldsymbol{A}} &= \frac{\det{\boldsymbol{A}}}{4\pi}
\int_{\lVert \underline{\xi} \rVert=1}
\frac{\underline{\xi}\stackrel{s}{\otimes}\boldsymbol{1}\stackrel{s}{\otimes}\underline{\xi}}
{\lVert \boldsymbol{A}\cdot\underline{\xi} \rVert^3}\,\mathrm{d} S_{\xi}
=
\frac{1}{4\pi}
\int_{\lVert \underline{\zeta} \rVert=1}
\frac{%
(\boldsymbol{A}^{-1}\cdot\underline{\zeta})
\stackrel{s}{\otimes}
\boldsymbol{1}
\stackrel{s}{\otimes}
(\boldsymbol{A}^{-1}\cdot\underline{\zeta})
}{\lVert \boldsymbol{A}^{-1}\cdot\underline{\zeta} \rVert^2}
\,\mathrm{d} S_{\zeta}\\
&=
\frac{\boldsymbol{1}\stackrel{s}{\boxtimes}\boldsymbol{I}^{\boldsymbol{A}}+\boldsymbol{I}^{\boldsymbol{A}}\stackrel{s}{\boxtimes}\boldsymbol{1}}{2}
\end{align}$$

## Proof of the change of variable in ([eq]) (click for details)

The proof detailed here is an intrinsic alternative (see [@barthelemyIJSS2016]) to the reasoning in components presented in [@mura1987].

The change of variable in ([eq]) is a bijective transformation of the unit sphere into itself
$$
\underline{\zeta}\mapsto\underline{\xi}=\frac{\boldsymbol{A}^{-1}⋅\underline{\zeta}}{\lVert \boldsymbol{A}^{-1}⋅\underline{\zeta} \rVert}
\;\Leftrightarrow\;
\underline{\xi}\mapsto\underline{\zeta}=\frac{\boldsymbol{A}⋅\underline{\xi}}{\lVert \boldsymbol{A}⋅\underline{\xi} \rVert}
$$

The delicate point of the proof is the following result on surface elements
$$
\,\mathrm{d} S_{\zeta}
=\frac{\det{\boldsymbol{A}}}{\lVert \boldsymbol{A}\cdot\underline{\xi} \rVert^3}\,
\,\mathrm{d} S_{\xi}
\;\Leftrightarrow\;
\,\mathrm{d} S_{\xi}
=\frac{\det{\boldsymbol{A}^{-1}}}{\lVert \boldsymbol{A}^{-1}\cdot\underline{\zeta} \rVert^3}\,
\,\mathrm{d} S_{\zeta}
$$

It is obvious that $\underline{\xi}$ and $\underline{\zeta}$ play symmetrical role by inverting $\boldsymbol{A}$.

Let's start by differentiating the transformation relationship yielding
$$
\,\mathrm{d} \underline{\xi}=\left(\boldsymbol{1}-\underline{\xi}\otimes\underline{\xi}\right)\cdot
\frac{\boldsymbol{A}^{-1}}{\lVert \boldsymbol{A}^{-1}\cdot\underline{\zeta} \rVert}\cdot\,\mathrm{d}\underline{\zeta}
$$
This transformation gradient can be decomposed as the tangent invertible application $\underline{u}\mapsto\frac{\boldsymbol{A}^{-1}}{\lVert \boldsymbol{A}^{-1}\cdot\underline{\zeta} \rVert}\cdot\underline{u}$ followed by an orthogonal projection over the plane of normal $\underline{\xi}$.

The end of the proof takes advantage of the following remarks:

- the transformation of a surface vector $\,\mathrm{d}\underline{S}$ to $\,\mathrm{d}\underline{s}$ by the gradient transformation $\underline{X}\mapsto\,\mathrm{d}\underline{x}=\boldsymbol{F}⋅\,\mathrm{d}\underline{X}$ is $\,\mathrm{d}\underline{s}=(\det{\boldsymbol{F}})\,{}^{t}{\boldsymbol{F}}^{-1}⋅\,\mathrm{d}\underline{S}$

- as $\underline{\xi}$ and $\underline{\zeta}$ belong to unit spheres, $\,\mathrm{d} \underline{S}_{\zeta}=\,\mathrm{d} S_{\zeta}\,\underline{\zeta}$ and $\,\mathrm{d} \underline{S}_{\xi}=\,\mathrm{d} S_{\xi}\,\underline{\xi}$

- the projection of material vectors over the plane of normal $\underline{\xi}$ becomes a projection on the line directed by $\underline{\xi}$ for surface vectors.

It results that
$$
\,\mathrm{d} S_{\xi}=\underline{\xi}\cdot\left(\left(\frac{\det{\boldsymbol{A}^{-1}}}{\lVert \boldsymbol{A}^{-1}\cdot\underline{\zeta} \rVert^3}\right)\,
\left(\lVert \boldsymbol{A}^{-1}\cdot\underline{\zeta} \rVert{}^{t}{\boldsymbol{A}}\right)\right)\cdot\underline{\zeta}
\,\mathrm{d} S_{\zeta}
=
\frac{\det{\boldsymbol{A}^{-1}}}{\lVert \boldsymbol{A}^{-1}\cdot\underline{\zeta} \rVert^2}\,
\underline{\zeta}\cdot\boldsymbol{A}\cdot\underline{\xi}
\,\mathrm{d} S_{\zeta}
$$
which ends the proof by observing that $\underline{\zeta}\cdot\boldsymbol{A}\cdot\underline{\xi}=\frac{\lVert \underline{\zeta} \rVert^2}{\lVert \boldsymbol{A}^{-1}\cdot\underline{\zeta} \rVert}=\frac{1}{\lVert \boldsymbol{A}^{-1}\cdot\underline{\zeta} \rVert}$.

By symmetry, it is obvious that $\boldsymbol{A}$ ([eq]) and $\boldsymbol{I}^{\boldsymbol{A}}$ are simultaneously diagonalizable i.e. they share the same eigenvectors:
$$
\boldsymbol{I}^{\boldsymbol{A}}=I^{\boldsymbol{A}}_1\,\underline{e}^{\boldsymbol{A}}_1\otimes\underline{e}^{\boldsymbol{A}}_1 + I^{\boldsymbol{A}}_2\,\underline{e}^{\boldsymbol{A}}_2\otimes\underline{e}^{\boldsymbol{A}}_2 + I^{\boldsymbol{A}}_3\,\underline{e}^{\boldsymbol{A}}_3\otimes\underline{e}^{\boldsymbol{A}}_3 = \sum_{i=1}^3 I^{\boldsymbol{A}}_i\,\underline{e}^{\boldsymbol{A}}_i\otimes\underline{e}^{\boldsymbol{A}}_i
$$

The components $I^{\boldsymbol{A}}_i$ can be obtained by identification with expressions derived from the Newtonian potential ([@kellogg1929], [@eshelby1957], [@parnell2016]). They are gathered in [tbl] or recalled a little later in a generic form ([@barthelemyIJSS2016], [@barthelemyIJES2020_hilltrans]).

## Hill polarization tensor in elasticity

### General expression

A general expression of the elastic polarization tensor is derived in [@willis1977] (see also [@mura1987])
$$\begin{align}
\mathbb{P}(\boldsymbol{A},\mathbb{C})&=\frac{1}{4\pi}
\int_{\lVert \underline{\zeta} \rVert=1}
(\boldsymbol{A}^{-1}\cdot\underline{\zeta})\stackrel{s}{\otimes}
\Big((\boldsymbol{A}^{-1}\cdot\underline{\zeta})\cdot\mathbb{C}
\cdot(\boldsymbol{A}^{-1}\cdot\underline{\zeta})\Big)^{-1}
\stackrel{s}{\otimes}(\boldsymbol{A}^{-1}\cdot\underline{\zeta})
\,\mathrm{d} S_\zeta\\
&=
\frac{\det{\boldsymbol{A}}}{4\pi}
\int_{\lVert \underline{\xi} \rVert=1}
\frac{\underline{\xi}\stackrel{s}{\otimes}
(\underline{\xi}\cdot\mathbb{C}
\cdot\underline{\xi})^{-1}
\stackrel{s}{\otimes}\underline{\xi}}{\lVert \boldsymbol{A}\cdot\underline{\xi} \rVert^3}
\,\mathrm{d} S_{\xi}
\end{align}$$
where $\mathbb{C}$ is the stiffness tensor of the reference medium.

When $\mathbb{C}$ is arbitrarily anisotropic, it is necessary to resort to numerical cubature to estimate $\mathbb{P}$ as proposed in [@ghahremani1977], [@gavazzi1990] or [@masson2008]. However in some cases of anisotropy, analytical solutions are available ([@withers1989], [@suvorov2002]). Note that the transformation technique ([@pouya2000], [@pouya2006]) allows to extend the set of available anisotropies leading to analytical Hill tensors [@barthelemyIJES2020_hilltrans]. The case of isotropic matrix is particularly developed in the next section.

### Isotropic matrix

In this section, the matrix is assumed isotropic so that its stiffness tensor writes by means of a bulk $k$ and shear $\mu$ or Lamé $\lambda$ and $\mu$ moduli or even Young modulus $E$ and Poisson ratio $\nu$ with $k=\frac{E}{3(1-2\nu)}$ and $\mu=\frac{E}{2(1+\nu)}$.
$$
\mathbb{C} = 3k\mathbb{J}+2\mu\mathbb{K} =  3\lambda\mathbb{I}+2\mu\mathbb{K}
$$

Introducing ([eq]) in ([eq]) leads to after some algebra
$$
\mathbb{P}(\boldsymbol{A},3\lambda\mathbb{I}+2\mu\mathbb{K})=
\frac{1}{\lambda+2\,\mu}
\mathbb{U}^{\boldsymbol{A}}
+\frac{1}{\mu}(\mathbb{V}^{\boldsymbol{A}}-\mathbb{U}^{\boldsymbol{A}})
$$
where the tensors $\mathbb{U}^{\boldsymbol{A}}$ and $\mathbb{V}^{\boldsymbol{A}}$, depending only on the ellipsoidal
tensor $\boldsymbol{A}$ of ([eq]), are given by ([eq]) and ([eq]).
Note that the writing ([eq]) puts well in evidence the contributions of the shape and orientation of the ellipsoid on the one hand and the moduli of the reference medium on the other hand.

## Hill polarization tensor in conductivity

### General expression

The conductivity polarization tensor related to a conductivity tensor $\boldsymbol{K}$ and an ellipsoid defined by $\boldsymbol{A}$ is also expressed in [@willis1977] under the form
$$\begin{align}
\boldsymbol{P}(\boldsymbol{A},\boldsymbol{K})&=\frac{1}{4\pi}
\int_{\lVert \underline{\zeta} \rVert=1}
\frac{(\boldsymbol{A}^{-1}\cdot\underline{\zeta})\stackrel{s}{\otimes}(\boldsymbol{A}^{-1}\cdot\underline{\zeta})}
{(\boldsymbol{A}^{-1}\cdot\underline{\zeta})\cdot\boldsymbol{K}
\cdot(\boldsymbol{A}^{-1}\cdot\underline{\zeta})}
\,\mathrm{d} S_\zeta\\
&=
\frac{\det{\boldsymbol{A}}}{4\pi}
\int_{\lVert \underline{\xi} \rVert=1}
\frac{\underline{\xi}\otimes\underline{\xi}}{(\underline{\xi}\cdot\boldsymbol{K}\cdot\underline{\xi})\,\lVert \boldsymbol{A}\cdot\underline{\xi} \rVert^3}
\,\mathrm{d} S_{\xi}
\end{align}$$

### Isotropic matrix

The case of an isotropic matrix $\boldsymbol{K}=K\,\boldsymbol{1}$ is straightforward:
$$
\boldsymbol{P}(\boldsymbol{A},K\,\boldsymbol{1})=
\frac{1}{K}\,
\left(
\frac{\det{\boldsymbol{A}}}{4\pi}
\int_{\lVert \underline{\xi} \rVert=1}
\frac{\underline{\xi}\otimes\underline{\xi}}{\lVert \boldsymbol{A}\cdot\underline{\xi} \rVert^3}
\,\mathrm{d} S_{\xi}\right)
=
\frac{\boldsymbol{I}^{\boldsymbol{A}}}{K}
$$
with $\boldsymbol{I}^{\boldsymbol{A}}$ defined in ([eq]) and derived in ([eq]) and [tbl].

### Arbitrarily anisotropic matrix

The complete solution of the second-order Hill polarization tensor in a matrix of arbitrary has been derived in [@barthelemyTIPM2009] from a reasoning based on Green solution. However a more direct solution, based on the idea of transformation, has been detailed in [@giraudMOM2019] and is recalled here. It first relies on the idea of square-root of the conductivity tensor. As a positive definite symmetric second-order tensor, $\boldsymbol{K}$ can be diagonalized in an orthonormal frame $\left(\boldsymbol{e}^K_i\right)_{i=1,2,3}$ with strictly positive eigenvalues $\left(K_i\right)_{i=1,2,3}$ such that
$$
\boldsymbol{K}=K_1 \underline{e}^{\boldsymbol{K}}_1\otimes\underline{e}^{\boldsymbol{K}}_1 + K_2 \underline{e}^{\boldsymbol{K}}_2\otimes\underline{e}^{\boldsymbol{K}}_2 + K_3 \underline{e}^{\boldsymbol{K}}_3\otimes\underline{e}^{\boldsymbol{K}}_3
$$
Its square-root is then defined by
$$
\boldsymbol{K}^{½}=\sqrt{K_1} \underline{e}^{\boldsymbol{K}}_1\otimes\underline{e}^{\boldsymbol{K}}_1 + \sqrt{K_2} \underline{e}^{\boldsymbol{K}}_2\otimes\underline{e}^{\boldsymbol{K}}_2 + \sqrt{K_3} \underline{e}^{\boldsymbol{K}}_3\otimes\underline{e}^{\boldsymbol{K}}_3
$$
of inverse
$$
\boldsymbol{K}^{-½}=\left(\boldsymbol{K}^{½}\right)^{-1}=\frac{1}{\sqrt{K_1}} \underline{e}^{\boldsymbol{K}}_1\otimes\underline{e}^{\boldsymbol{K}}_1 + \frac{1}{\sqrt{K_2}} \underline{e}^{\boldsymbol{K}}_2\otimes\underline{e}^{\boldsymbol{K}}_2 + \frac{1}{\sqrt{K_3}} \underline{e}^{\boldsymbol{K}}_3\otimes\underline{e}^{\boldsymbol{K}}_3
$$

It follows that ([eq]) rewrites
$$\begin{align}
\boldsymbol{P}(\boldsymbol{A},\boldsymbol{K})&=\frac{1}{4\pi}
\int_{\lVert \underline{\zeta} \rVert=1}
