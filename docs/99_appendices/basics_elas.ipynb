# Basic problem of elasticity homogenization

## System of equations

Consider a representative volume element (RVE) $\Omega$ composed of a heterogeneous material. Neglecting body forces in a problem posed at the scale of a RVE is consistent with the fact that the order of magnitude of mechanical effects induced by body forces is in general much lower than that of the macroscopic strain $\boldsymbol{E}$ or stress $\boldsymbol{\Sigma}$ effects accounting for interactions with particles surrounding the RVE (see [@dormieux2006]). The hypothesis of quasi-static equilibrium is also invoked here to write the balance law involving the Cauchy stress field $\boldsymbol{\sigma}$
$$
\underline{\operatorname{div}}{\boldsymbol{\sigma}}=\underline{0} \quad(\Omega)
$$

In the sequel, the small perturbation hypothesis is adopted so that the strain field $\boldsymbol{\varepsilon}$ derives from the displacement one $\underline{u}$ as the symmetrical part of its gradient
$$
\boldsymbol{\varepsilon}=\frac{\boldsymbol{\operatorname{grad}}{\underline{u}}+{}^{t}{\boldsymbol{\operatorname{grad}}{\underline{u}}}}{2} \quad(\Omega)
$$

In the framework of random media homogenization, two types of conditions applied at the boundary $\partial\Omega$ of a RVE $\Omega$ are usually considered:

- *homogeneous strain boundary conditions* corresponding to prescribed displacements $\underline{u}^g$ at $\partial\Omega$
$$
\underline{u}^g=\boldsymbol{E}\cdot\underline{x} \quad(\partial\Omega)
$$
It is noticeable that in this case the divergence theorem implies the following relationship between the microscopic and macroscopic strain tensors
$$
<\boldsymbol{\varepsilon}>_{\Omega}=\frac{1}{\lvert \Omega \rvert}\int_\Omega \boldsymbol{\varepsilon}\,\mathrm{d} \Omega
=\frac{1}{\lvert \Omega \rvert}\int_{\partial\Omega} \underline{u}\stackrel{s}{\otimes}\underline{n} \,\mathrm{d} S
=\boldsymbol{E}
$$
where the spatial average over a domain $\omega$ is denoted by $<\bullet>_\omega$ and $\underline{n}$ is the unit outward normal at the boundary.
The macroscopic stress tensor is then simply defined as the average
$$
\boldsymbol{\Sigma}=
<\boldsymbol{\sigma}>_{\Omega}=\frac{1}{\lvert \Omega \rvert}\int_\Omega \boldsymbol{\sigma}\,\mathrm{d} \Omega
$$

- *homogeneous stress boundary conditions* corresponding to prescribed surface tractions $\underline{T}^g$ at $\partial\Omega$
$$
\underline{T}^g=\boldsymbol{\Sigma}\cdot\underline{n} \quad(\partial\Omega)
$$
Now owing to the remarkable identity $(x_i\sigma_{jk})_{,k}=\sigma_{ij}$ resulting from ([eq]) and the symmetry of $\boldsymbol{\sigma}$, the relationship between the microscopic and macroscopic stress is ensured by the divergence theorem
$$
<\boldsymbol{\sigma}>_{\Omega}=\frac{1}{\lvert \Omega \rvert}\int_\Omega \boldsymbol{\sigma}\,\mathrm{d} \Omega
=\frac{1}{\lvert \Omega \rvert}\int_{\partial\Omega} \underline{x}\stackrel{s}{\otimes}(\boldsymbol{\sigma}\cdot\underline{n}) \,\mathrm{d} S
=\boldsymbol{\Sigma}
$$
The macroscopic stress tensor is then simply defined as the average
$$
\boldsymbol{E}=
<\boldsymbol{\varepsilon}>_{\Omega}=\frac{1}{\lvert \Omega \rvert}\int_\Omega \boldsymbol{\varepsilon}\,\mathrm{d} \Omega
$$

## Hill lemma

Note that whatever the choice of boundary conditions between ([eq]) and ([eq]), the consistency between the microscopic and macroscopic works is ensured by
$$
<\boldsymbol{\sigma}:\boldsymbol{\varepsilon}>_{\Omega}
=\frac{1}{\lvert \Omega \rvert}\int_\Omega \boldsymbol{\sigma}:\boldsymbol{\varepsilon}\,\mathrm{d} \Omega
=\frac{1}{\lvert \Omega \rvert}\int_{\partial\Omega} \underline{u}\cdot\boldsymbol{\sigma}\cdot\underline{n} \,\mathrm{d} S
=\boldsymbol{\Sigma}:\boldsymbol{E}
$$
which results from the application of the divergence theorem to $(u_i\sigma_{ij})_{,k}=u_{i,j}\sigma_{ij}=\varepsilon_{ij}\sigma_{ij}$.

The set of equations defining the problem posed on the RVE is finally completed by the local constitutive law relating the strain and stress fields. The hypothesis of linear elasticity is adopted in this part so that
$$
\boldsymbol{\sigma}=\mathbb{c}:\boldsymbol{\varepsilon} \quad(\Omega)
$$
where $\mathbb{c}(\underline{x})$ denotes the heterogeneous (positive definite fourth-order) stiffness tensor field satisfying the conditions of minor ($c_{jikl}=c_{ijlk}=c_{ijkl}$) and major ($c_{klij}=c_{ijkl}$) symmetries. ## Macroscopic stiffness or compliance tensors

Whatever the boundary condition of homogeneous strain or stress type ([eq]) or ([eq]), the linearity of the problem allows to invoke the existence of concentration tensors relating the microscopic strain $\boldsymbol{\varepsilon}$ and stress $\boldsymbol{\sigma}$ fields to the macroscopic strain $\boldsymbol{E}$ or stress $\boldsymbol{\Sigma}$ tensors
$$\begin{align}
\boldsymbol{\varepsilon} & =  \mathbb{A}_E:\boldsymbol{E} \\
\boldsymbol{\sigma} & =  \mathbb{B}_E:\boldsymbol{E} \quad\textrm{with}\quad \mathbb{B}_E=\mathbb{c}:\mathbb{A}_E \\
\boldsymbol{\sigma} & =  \mathbb{B}_\Sigma:\boldsymbol{\Sigma} \\
\boldsymbol{\varepsilon} & =  \mathbb{A}_\Sigma:\boldsymbol{\Sigma} \quad\textrm{with}\quad \mathbb{A}_\Sigma=\mathbb{s}:\mathbb{B}_\Sigma 
\end{align}$$
