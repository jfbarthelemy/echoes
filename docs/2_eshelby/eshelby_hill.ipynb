# Eshelby and Hill polarization tensors

## Objectives

This tutorial provides the instructions to build Eshelby and Hill polarization tensors related to an ellipsoidal shape embedded in an infinite matrix of arbitrary anisotropy, both in **elasticity** (4th-order tensors) and in **transport** problems such as conductivity or diffusion (2nd-order tensors).

## Imports

````python
import numpy as np
from echoes import *
import math

np.set_printoptions(precision=8, suppress=True)
# to display only 8 significant digits of array components
````

## The Eshelby inclusion problem in elasticity [@eshelby1957]

Let the studied domain be the whole $\mathbb{R}^3$ space here filled with a linear elastic material of uniform stiffness tensor $\mathbb{C}$. An ellipsoid centered at the origin of the domain is defined by the equation

$$
   \underline{x}\in\mathcal{E}_{\boldsymbol{A}}
\quad\Leftrightarrow\quad   
\underline{x}\cdot({}^{t}{\boldsymbol{A}}\cdot\boldsymbol{A})^{-1}\cdot\underline{x}\leq 1
$$
where $\boldsymbol{A}$ is an invertible second-order tensor so that ${}^{t}{\boldsymbol{A}}\cdot\boldsymbol{A}$ is a positive definite symmetric tensor associated to 3 radii (eigenvalues $a\geq b \geq c$ possibly written $\rho_1 \geq \rho_2 \geq \rho_3$ for convenience) and 3 angles (orientation of the frame of orthonormal eigenvectors $\underline{e}^{\boldsymbol{A}}_1, \underline{e}^{\boldsymbol{A}}_2, \underline{e}^{\boldsymbol{A}}_3$)
$$
{}^{t}{\boldsymbol{A}}\cdot\boldsymbol{A}=a^2 \underline{e}^{\boldsymbol{A}}_1\otimes\underline{e}^{\boldsymbol{A}}_1 + b^2 \underline{e}^{\boldsymbol{A}}_2\otimes\underline{e}^{\boldsymbol{A}}_2 + c^2 \underline{e}^{\boldsymbol{A}}_3\otimes\underline{e}^{\boldsymbol{A}}_3 = \sum_{i=1}^3 \rho_i^2 \underline{e}^{\boldsymbol{A}}_i\otimes\underline{e}^{\boldsymbol{A}}_i
$$

A polarization field is introduced as a second-order symmetric tensor field $\boldsymbol{\tau}$ such that the constitutive law of the medium is given by

$$
    \boldsymbol{\sigma}(\underline{x})=\mathbb{C}:\boldsymbol{\varepsilon}(\underline{x})+\boldsymbol{\tau}(\underline{x})\quad\textrm{where}\quad
    \boldsymbol{\tau}(\underline{x})=
    \left\{
    \begin{array}{ll}
    \boldsymbol{0}&\textrm{ if }\underline{x} \notin \mathcal{E}_{\boldsymbol{A}}\\
    \boldsymbol{\tau}\textrm{ (uniform)}&\textrm{ if }\underline{x} \in \mathcal{E}_{\boldsymbol{A}}
    \end{array}
    \right.
$$

In addition the displacement tends towards $\underline{0}$ at infinity.

![](../img/eshelbypbincl.png){width=30%}

Eshelby inclusion problem

The important result derived in [@eshelby1957] is that the strain tensor field solution to this problem is uniform within the ellipsoidal domain (not outside!) and writes

$$
   \forall \underline{x} \in \mathcal{E}_{\boldsymbol{A}},\quad \boldsymbol{\varepsilon}(\underline{x})=-\mathbb{P}:\boldsymbol{\tau}\quad\textrm{(uniform)}
$$
where $\mathbb{P}$, called the Hill polarization tensor, depends on the shape and orientation of the ellipsoid (i.e. $\boldsymbol{A}$ up to a multiplicative constant) and on the matrix behavior (i.e. $\mathbb{C}$), in other words $\mathbb{P}=\mathbb{P}(\boldsymbol{A},\mathbb{C})$. See in [§] explicit expressions to calculate the Hill polarization tensor.

The polarization problem can be alternatively posed thanks to an eigenstrain tensor $\boldsymbol{\varepsilon}^*$, uniform in the ellipsoid, such that $\boldsymbol{\varepsilon}^*=-\mathbb{C}^{-1}:\boldsymbol{\tau}$. It follows that ([eq]) rewrites

$$
    \boldsymbol{\sigma}(\underline{x})=\mathbb{C}:(\boldsymbol{\varepsilon}(\underline{x})-\boldsymbol{\varepsilon}^*(\underline{x}))\quad\textrm{where}\quad
    \boldsymbol{\varepsilon}^*(\underline{x})=
    \left\{
    \begin{array}{ll}
    \boldsymbol{0}&\textrm{ if }\underline{x} \notin \mathcal{E}_{\boldsymbol{A}}\\
    \boldsymbol{\varepsilon}^*\textrm{ (uniform)}&\textrm{ if }\underline{x} \in \mathcal{E}_{\boldsymbol{A}}
    \end{array}
    \right.
$$
and the Eshelby tensor $\mathbb{S}$ is introduced by rewriting ([eq]) under the form
$$
   \forall \underline{x} \in \mathcal{E}_{\boldsymbol{A}},\quad \boldsymbol{\varepsilon}(\underline{x})=\mathbb{S}:\boldsymbol{\varepsilon}^*\quad\textrm{where}\quad\mathbb{S}=\mathbb{P}:\mathbb{C}
$$

A straightforward consequence of the uniformity of the strain state wihin the ellipsoid is the uniformity of the stress state. A dual point of view of the expression ([eq]) consists in relating the uniform stress within the ellipsoid to the eigenstrain introduced in ([eq]). Using ([eq]) in ([eq]) allows to exhibit the dual counterpart of ([eq]) involving the so-called second Hill polarization tensor $\mathbb{Q}$
$$
   \forall \underline{x} \in \mathcal{E}_{\boldsymbol{A}},\quad \boldsymbol{\sigma}(\underline{x})=-\mathbb{Q}:\boldsymbol{\varepsilon}^*\quad\textrm{with}\quad\mathbb{Q}=\mathbb{C}-\mathbb{C}:\mathbb{P}:\mathbb{C}
$$

## Definition of the `ellipsoidal` shape

An `ellipsoidal` shape intended for use in the construction of $\mathbb{P}$ can be built by means of the following constructor

`shape = ellipsoidal(a, b, c, θ=0., ϕ=0., ψ=0.) # default values 0 for the angles`

where $a, b, c$ are the radii and $\theta, \phi, \psi$ the Euler angles already presented in [fig]

````python
π = math.pi
a, b, c = 10., 5., 2.
θ, ϕ, ψ = π/3, π/4, π/5
print(ellipsoidal(a, b, c, θ, ϕ, ψ))
````

## Warning

If the radii are not entered in decreasing order, the constructor reorders them and recomputes the angles accordingly.

See below how the constructor reorders the radii and recomputes the angles accordingly

When two radii are equal the ellipsoidal shape becomes a spheroidal one which can also be directly defined by

`shape = spheroidal(ω, θ=0., ϕ=0.) # default values 0 for the angles`

where $\omega$ is the aspect ratio ($\omega \leq 1$ for an oblate shape and $\omega \geq 1$ for a prolate shape) (see [fig]).

Finally if the three radii are equal, the shape becomes spherical

`shape = spherical`

This explains why the spheroidal shape is defined only with an aspect ratio and the sphere is not defined with a radius. ## Calculation of Hill and Eshelby tensors for an isotropic matrix

The Hill, second Hill and Eshelby tensors are simply calculated by

- `P = hill(shape, C)`
- `Q = hill_dual(shape, C)`
- `S = eshelby(shape, C)`

where `shape` is an ellipsoidal shape as previously defined and `C` is an isotropic tensor.

Note that an isotropic tensor `C` simply allows to resort to analytical expressions as seen in [§]. The anisotropic cases are also available with numerical algorithms as will be seen below.

## Warning

The functions `hill` and `eshelby` return $6×6$ matrices i.e. of type `numpy.ndarray` (not `tensor`). It is however possible to build a `tensor` object from the matrix for the Hill tensor $\mathbb{P}$ since the latter is always positive definite. This is not in general the case for the Eshelby tensor $\mathbb{S}$.

````python
a, b, c = 10., 5., 2.
θ, ϕ, ψ = π/3, π/4, π/5
k, µ = 72., 32. 
shape = ellipsoidal(a, b, c, θ, ϕ, ψ)
C = stiff_kmu(k, µ)
P = hill(shape, C)
print("P =\n", P)
````

A `tensor` object can be constructed

````python
print("tensor P =\n", tensor(P, optiangles=True))
````

````python
S = eshelby(shape, C)
print("S =\n", S)
````

## Calculation of Hill and Eshelby tensors for an arbitrary anisotropic matrix 

The Hill, second Hill and Eshelby tensors are numerically calculated by

- `P = hill(shape, C, algo=NUMINT, epsroots=1.e-4, epsabs=1.e-4, epsrel=1.e-4, maxnb=100000)`
- `Q = hill_dual(shape, C, algo=NUMINT, epsroots=1.e-4, epsabs=1.e-4, epsrel=1.e-4, maxnb=100000)`
- `S = eshelby(shape, C, algo=NUMINT, epsroots=1.e-4, epsabs=1.e-4, epsrel=1.e-4, maxnb=100000)`

where the keyword argument `algo` can be either `NUMINT` or `RESIDUES` (default: `NUMINT`).

- `NUMINT` based on the cubature DECUHR algorithm provided in [@espelid1994] applied on the integral formulation ([eq]) over the unit sphere
    - `epsabs` is the tolerance of absolute error (default: 1.e-4)
    - `epsrel` is the tolerance of relative error (default: 1.e-4)
    - `maxnb` is the maximum number of subdivisions (default: 100000)
    
  For more information about these parameters see [@espelid1994]. 

- `RESIDUES` based on the residue Cauchy theorem to transform the 2D cubature over the unit sphere ([eq]) into a 1D quadrature as demonstrated in [@masson2008]
    - `epsroots` is the tolerance for the criterion of multiple poles, i.e. if the distance between two poles of the integrand function is under this value they are considered as a unique multiple pole (default: 1.e-4)
    - `epsabs` is the tolerance of absolute error for the 1D quadrature (default: 1.e-4)
    - `epsrel` is the tolerance of relative error for the 1D quadrature (default: 1.e-4)
    - `maxnb` is the maximum number of subdivisions for the 1D quadrature (default: 100000)

## Warning

The `RESIDUES` algorithm is much faster than the `NUMINT` one in case of full anisotropy. Nevertheless the latter proves to be more robust in the neighborhood of symmetry axes where the presence of close poles may induce numerical errors in the application of the Cauchy residue theorem. This is why `NUMINT` is chosen by default.

````python
param = [ 2.66011, 1.26432, 0.662772, 1.9402, 1.54905, 1.10384, 3.6072, 1.78964, 2.0247, 1.2701, 1.1089, 2.743, 1.3367, 1.2962, 0.897632, 4.42684, 2.05632, 1.52686, 3.54431, 1.3445, 1.99356 ]
C = tensor(param)
print("Hill tensor using NUMINT algorithm\n", hill(shape, C, algo=NUMINT),"\n")
````

````python
print("Hill tensor using RESIDUES algorithm\n", hill(shape, C, algo=RESIDUES))
````

See below how a too large `epsroots` parameter can degrade the precision of the result

````python
param = [3.,4.,math.sqrt(2),2.,1.]
np.random.seed(1948)
α = 1.e-2*np.random.rand(6, 6)
C = tensor(param)+tensor((α+α.T)/2) ; print("C =\n",C)
print("P(NUMINT) =\n", hill(shape, C, algo=NUMINT),"\n")
````

````python
for ϵ in [1.e-4, 1.e-2, 1.e-1]:
    print(f"ϵ={ϵ}\n", "  P =\n", hill(shape, C, algo=RESIDUES, epsroots=ϵ),"\n")
````

## 2nd-order Hill and Eshelby tensors for transport properties

The Eshelby framework extends verbatim to **transport problems** (heat conduction, mass diffusion, electric conduction…). The whole space is now filled with a uniform **conductivity tensor** $\boldsymbol{K}$ (a positive-definite 2nd-order symmetric tensor). Denoting $\boldsymbol{j}=-\boldsymbol{q}$ the flux vector (so that $\boldsymbol{j}=\boldsymbol{K}\cdot\text{grad}\,T$ in the absence of polarization), a uniform polarization $\boldsymbol{\tau}$ is prescribed inside the ellipsoid:

$$
\boldsymbol{j}(\underline{x}) = \boldsymbol{K}\cdot\text{grad}\,T(\underline{x}) + \boldsymbol{\tau}(\underline{x})
\quad\textrm{with}\quad
\text{grad}\,T(\underline{x}) \xrightarrow{\lVert \underline{x} \rVert\to\infty} \underline{0}
$$

The key result is that the temperature gradient inside the ellipsoid is **uniform**:

$$
\forall\underline{x}\in\mathcal{E}_{\boldsymbol{A}},\quad
\text{grad}\,T(\underline{x}) = -\boldsymbol{P}(\boldsymbol{A},\boldsymbol{K})\cdot\boldsymbol{\tau}
$$

where $\boldsymbol{P}(\boldsymbol{A},\boldsymbol{K})$ is now a **2nd-order symmetric tensor** (3×3 matrix). The conductivity Eshelby tensor is accordingly $\boldsymbol{S} = \boldsymbol{P}\cdot\boldsymbol{K}$. Explicit formulae are given in [§]; the key results recalled here are:

- **Isotropic matrix** $\boldsymbol{K}=K\,\boldsymbol{1}$:

$$
\boldsymbol{P}(\boldsymbol{A}, K\,\boldsymbol{1}) = \frac{\boldsymbol{I}^{\boldsymbol{A}}}{K}
$$

where $\boldsymbol{I}^{\boldsymbol{A}}$ is the Newton potential tensor defined in [eq].

- **Sphere** ($\boldsymbol{A}=\boldsymbol{1}$) in an isotropic matrix, $\boldsymbol{I}^{\boldsymbol{1}}=\frac{1}{3}\,\boldsymbol{1}$:

$$
\boldsymbol{P}(\boldsymbol{1}, K\,\boldsymbol{1}) = \frac{1}{3\,K}\,\boldsymbol{1}, \qquad
\boldsymbol{S}(\boldsymbol{1}, K\,\boldsymbol{1}) = \frac{1}{3}\,\boldsymbol{1}
$$

Note that the 2nd-order Eshelby tensor for a sphere is independent of $K$, just as its 4th-order elastic counterpart is independent of the inclusion stiffness.

### API

The library selects the 2nd- or 4th-order algorithm automatically from the order of the `ref` tensor passed to `hill` or `eshelby`. The calls are therefore **identical** to the elastic case:

```
P = hill(shape, K)
S = eshelby(shape, K)
```

where `K` is a 2nd-order tensor (e.g. `k * tId2` for an isotropic medium of conductivity $k$, or `tensor([k1, k2, k3], angles=[θ, φ, ψ])` for an orthotropic one). Both functions return a 3×3 `numpy.ndarray`.

### Isotropic matrix

````python
K_val = 2.5
K_iso = K_val * tId2     # isotropic conductivity  K = 2.5

print("P (sphere, isotropic) =\n", hill(spherical, K_iso), "\n")
print("S (sphere, isotropic) =\n", eshelby(spherical, K_iso), "\n")

# Spheroidal inclusions in the same matrix
for ω in [0.1, 0.5, 1., 2.]:
    P2 = hill(spheroidal(ω), K_iso)
    print(f"P (spheroid ω={ω}, isotropic) =\n", P2)
````

### Anisotropic matrix

Unlike the 4th-order elastic case, the 2nd-order Hill tensor is **always computed analytically** for any matrix symmetry. The key is the transformation formula [eq] from [§]:

$$
\boldsymbol{P}(\boldsymbol{A},\boldsymbol{K}) = \boldsymbol{K}^{-\frac{1}{2}}\cdot\boldsymbol{I}^{\boldsymbol{A}\cdot\boldsymbol{K}^{-\frac{1}{2}}}\cdot\boldsymbol{K}^{-\frac{1}{2}}
$$

which reduces the problem to the computation of the Newton potential tensor $\boldsymbol{I}^{\tilde{\boldsymbol{A}}}$ of a fictitious isotropic-equivalent ellipsoid $\tilde{\boldsymbol{A}}=\boldsymbol{A}\cdot\boldsymbol{K}^{-\frac{1}{2}}$, for which closed-form expressions are available. No numerical integration is therefore needed.

````python
# Orthotropic conductivity tensor with principal values and orientation angles
K_aniso = tensor([3.2, 0.5, 1./(3.2*0.5)], angles=[π/3, π/4, π/5])
print("K (orthotropic) =\n", K_aniso, "\n")

shape2 = ellipsoidal(10., 5., 2., π/3, π/4, π/5)

P2 = hill(shape2, K_aniso)
print("P2 (anisotropic K) =\n", P2)
print("\nS2 (anisotropic K) =\n", eshelby(shape2, K_aniso))
````
